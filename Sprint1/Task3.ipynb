<h2 style="color:DodgerBlue">Название проекта:</h2>

Классы студентов с наследованием и полиморфизмом

### Вариант задания №2


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.

2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.

3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// ==========================================
// БАЗОВЫЙ КЛАСС
// ==========================================

public class Student
{
    // Свойства с геттерами и сеттерами
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }

    // Конструктор с использованием свойств (сеттеров)
    public Student(string name, int age, int course)
    {
        this.Name = name;     // Вызывается сеттер
        this.Age = age;       
        this.Course = course; 
    }

    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится по основной программе.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен по основным дисциплинам.");
    }

    // === ВЗАИМОДЕЙСТВИЕ 1: Универсальное для всех студентов ===
    public void StudyTogether(Student other)
    {
        Console.WriteLine($"{this.Name} и {other.Name} готовятся к экзамену вместе.");
    }
}

// ==========================================
// 3 ПРОИЗВОДНЫХ КЛАССА
// ==========================================

public class BachelorStudent : Student
{
    public string Specialty { get; set; }

    public BachelorStudent(string name, int age, int course, string specialty)
        : base(name, age, course)
    {
        this.Specialty = specialty; // Вызывается сеттер
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности: {Specialty}");
    }

    // === ВЗАИМОДЕЙСТВИЕ 2: Бакалавр просит помощи у магистра ===
    public void AskForHelp(MasterStudent master)
    {
        Console.WriteLine($"Бакалавр {this.Name} просит помощи у магистра {master.Name}.");
    }
}

public class MasterStudent : Student
{
    public string ScientificAdvisor { get; set; }

    public MasterStudent(string name, int age, int course, string scientificAdvisor)
        : base(name, age, course)
    {
        this.ScientificAdvisor = scientificAdvisor; // Вызывается сеттер
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}";
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен под наблюдением научного руководителя {ScientificAdvisor}");
    }

    // === ВЗАИМОДЕЙСТВИЕ 3: Магистр консультирует бакалавра ===
    public void Mentor(BachelorStudent bachelor)
    {
        Console.WriteLine($"Магистр {this.Name} консультирует бакалавра {bachelor.Name} по специальности {bachelor.Specialty}.");
    }
}

public class PhDStudent : Student
{
    public string DissertationTopic { get; set; }

    public PhDStudent(string name, int age, int course, string dissertationTopic)
        : base(name, age, course)
    {
        this.DissertationTopic = dissertationTopic; // Вызывается сеттер
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: {DissertationTopic}";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} исследует тему диссертации: {DissertationTopic}");
    }

    public void ConductResearch()
    {
        Console.WriteLine($"{Name} проводит научные исследования по теме: {DissertationTopic}");
    }

    // === ВЗАИМОДЕЙСТВИЕ 4: Аспирант делится знаниями с любым студентом ===
    public void ShareKnowledge(Student student)
    {
        Console.WriteLine($"Аспирант {this.Name} делится опытом о диссертации («{this.DissertationTopic}») со студентом {student.Name}.");
    }
}

// ==========================================
//  ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ
// ==========================================

Console.WriteLine("=== Демонстрация наследования и полиморфизма ===\n");

// 1. Создаем объекты (работают конструкторы с сеттерами)
BachelorStudent bachelor = new BachelorStudent("Иван", 20, 3, "Программная инженерия");
MasterStudent master = new MasterStudent("Елена", 24, 1, "Профессор Сидоров");
PhDStudent phd = new PhDStudent("Александр", 28, 4, "Машинное обучение в биоинформатике");

Console.WriteLine("--- Взаимодействие объектов ---");

// Взаимодействие 1: Универсальное (базовый класс)
bachelor.StudyTogether(master); 

// Взаимодействие 2 и 3: Специфичное между бакалавром и магистром
bachelor.AskForHelp(master);
master.Mentor(bachelor);

Console.WriteLine();

// Взаимодействие 4: Аспирант взаимодействует с бакалавром
phd.ShareKnowledge(bachelor);

// Взаимодействие 5: Аспирант взаимодействует с магистром
phd.ShareKnowledge(master);

Console.WriteLine("\n--- Информация о студентах (Полиморфизм) ---");
Student[] students = { bachelor, master, phd };

foreach (Student student in students)
{
    Console.WriteLine(student.GetInfo());
    student.Study();
    student.TakeExam();
    Console.WriteLine(new string('-', 50));
}

// 3. Демонстрация специфичных методов
Console.WriteLine("\n--- Специфичные методы ---");
phd.ConductResearch();
```
