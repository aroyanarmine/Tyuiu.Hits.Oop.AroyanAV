<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма. Добавить к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой. Реализовать простое, сложное и множественное наследование. Реализовать полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.

2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.

3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями

```
using System;
using System.Collections.Generic;

public interface IHasScholarship
{
    decimal Scholarship { get; set; }
    void ReceiveScholarship();
}

public interface ICanTeach
{
    void TeachClass();
}

public interface ICanPublish
{
    void PublishWork();
}

public class Student
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }
    public string UniversityName { get; set; }
    
    public string Email { get; set; }

    public Student(string name, int age, int course, string universityName)
    {
        Name = name;
        Age = age;
        Course = course;
        UniversityName = universityName;
    }

    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}, ВУЗ: {UniversityName}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится по основной программе.");
    }

    public virtual void Study(int hours)
    {
        Console.WriteLine($"{Name} учится {hours} часов в день.");
    }

    public virtual void Study(string subject)
    {
        Console.WriteLine($"{Name} изучает дисциплину: {subject}.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен по основным дисциплинам.");
    }

    public void AttendLecture()
    {
        Console.WriteLine($"{Name} посещает лекцию в {UniversityName}.");
    }

    public void StudyTogether(Student other)
    {
        Console.WriteLine($"{this.Name} и {other.Name} готовятся к экзамену вместе.");
    }

    public void SendEmail(string subject)
    {
        Console.WriteLine($"{Name} получил письмо на {Email}: «{subject}»");
    }
}

public class BachelorStudent : Student, IHasScholarship
{
    public string Specialty { get; set; }
    public decimal Scholarship { get; set; }
    
    public string InternshipCompany { get; set; }

    public BachelorStudent(string name, int age, int course, string universityName, string specialty, decimal scholarship)
        : base(name, age, course, universityName)
    {
        Specialty = specialty;
        Scholarship = scholarship;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}, Стипендия: {Scholarship:C}";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности: {Specialty}");
    }

    public override void Study(int hours)
    {
        Console.WriteLine($"{Name} изучает специальность {Specialty} по {hours} часа в день.");
    }

    public void ReceiveScholarship()
    {
        Console.WriteLine($"{Name} получил стипендию в размере {Scholarship:C}.");
    }

    public void AskForHelp(MasterStudent master)
    {
        Console.WriteLine($"Бакалавр {this.Name} просит помощи у магистра {master.Name}.");
    }

    public void DoInternship()
    {
        Console.WriteLine($"{Name} проходит практику в компании {InternshipCompany}.");
    }
}

public class MasterStudent : Student, ICanTeach, ICanPublish
{
    public string ScientificAdvisor { get; set; }
    public int PublicationCount { get; set; }
    
    public int ConferenceCount { get; set; }

    public MasterStudent(string name, int age, int course, string universityName, string scientificAdvisor, int publicationCount)
        : base(name, age, course, universityName)
    {
        ScientificAdvisor = scientificAdvisor;
        PublicationCount = publicationCount;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}, Публикаций: {PublicationCount}";
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен под наблюдением научного руководителя {ScientificAdvisor}");
    }

    public override void Study(string subject)
    {
        Console.WriteLine($"Магистр {Name} углублённо изучает {subject} и преподаёт его.");
    }

    public void TeachClass()
    {
        Console.WriteLine($"Магистр {Name} проводит семинар для младших курсов.");
    }

    public void TeachClass(string topic)
    {
        Console.WriteLine($"Магистр {Name} проводит семинар по теме: {topic}.");
    }

    public void PublishArticle()
    {
        PublicationCount++;
        Console.WriteLine($"{Name} опубликовал новую статью. Всего публикаций: {PublicationCount}");
    }

    public void Mentor(BachelorStudent bachelor)
    {
        Console.WriteLine($"Магистр {this.Name} консультирует бакалавра {bachelor.Name} по специальности {bachelor.Specialty}.");
    }

    public void AttendConference()
    {
        ConferenceCount++;
        Console.WriteLine($"{Name} выступил(а) на конференции. Всего участий: {ConferenceCount}");
    }

    void ICanPublish.PublishWork()
    {
        Console.WriteLine($"Магистр {Name} публикует статью в сборнике конференции.");
    }
}

public class PhDStudent : Student, IHasScholarship, ICanTeach, ICanPublish
{
    public string DissertationTopic { get; set; }
    public string GrantName { get; set; }
    public decimal Scholarship { get; set; }
    
    public int PatentCount { get; set; }

    public PhDStudent(string name, int age, int course, string universityName, string dissertationTopic, string grantName, decimal scholarship)
        : base(name, age, course, universityName)
    {
        DissertationTopic = dissertationTopic;
        GrantName = grantName;
        Scholarship = scholarship;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема: '{DissertationTopic}', Грант: {GrantName}, Стипендия: {Scholarship:C}";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} исследует тему диссертации: {DissertationTopic}");
    }

    public override void Study(int hours)
    {
        Console.WriteLine($"{Name} тратит {hours} часов в день на исследование темы: {DissertationTopic}");
    }

    public void ReceiveScholarship()
    {
        Console.WriteLine($"{Name} получает повышенную стипендию по гранту {GrantName}: {Scholarship:C}");
    }

    public void TeachClass()
    {
        Console.WriteLine($"Аспирант {Name} ведет лабораторные занятия.");
    }

    public void ApplyForGrant()
    {
        Console.WriteLine($"{Name} подает заявку на продление гранта '{GrantName}'.");
    }

    public void ShareKnowledge(Student student)
    {
        Console.WriteLine($"Аспирант {this.Name} делится опытом о диссертации («{this.DissertationTopic}») со студентом {student.Name}.");
    }

    public void FilePatent()
    {
        PatentCount++;
        Console.WriteLine($"{Name} подал(а) патент по теме диссертации. Всего патентов: {PatentCount}");
    }

    void ICanPublish.PublishWork()
    {
        Console.WriteLine($"Аспирант {Name} публикует статью в журнале Q1 по теме: {DissertationTopic}");
    }
}

public class SeniorPhDStudent : PhDStudent
{
    public bool IsDefendingSoon { get; set; }

    public SeniorPhDStudent(string name, int age, int course, string universityName, string dissertationTopic, string grantName, decimal scholarship, bool isDefendingSoon)
        : base(name, age, course, universityName, dissertationTopic, grantName, scholarship)
    {
        IsDefendingSoon = isDefendingSoon;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Защита скоро: {(IsDefendingSoon ? "Да" : "Нет")}";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} пишет заключительные главы диссертации: {DissertationTopic}");
    }

    public void PrepareDefense()
    {
        if (IsDefendingSoon)
            Console.WriteLine($"{Name} активно готовится к предзащите диссертации по теме: {DissertationTopic}");
        else
            Console.WriteLine($"{Name} пока собирает данные для будущей защиты.");
    }
}

public class StudentCollection<T> where T : Student
{
    private List<T> _students = new List<T>();

    public void Add(T student) { _students.Add(student); }
    public void Remove(T student) { _students.Remove(student); }
    public int Count { get { return _students.Count; } }

    public void ShowAllInfo()
    {
        foreach (var student in _students)
            Console.WriteLine(student.GetInfo());
    }

    public void AllStudy()
    {
        foreach (var student in _students)
            student.Study();
    }
}

public class ScholarshipService
{
    private readonly IHasScholarship _recipient; 

    public ScholarshipService(IHasScholarship recipient)
    {
        _recipient = recipient;
    }

    public void PayScholarship()
    {
        Console.WriteLine("Служба стипендий: начисляем выплату.");
        _recipient.ReceiveScholarship(); 
    }
}

public class TeachingService
{
    private readonly ICanTeach _teacher;

    public TeachingService(ICanTeach teacher)
    {
        _teacher = teacher;
    }

    public void ScheduleClass()
    {
        Console.WriteLine("Служба расписания: планируем занятие.");
        _teacher.TeachClass();
    }
}
```

```
Console.WriteLine("=== Индивидуальный проект: Иерархия Student ===\n");

// Создаём объекты; новые атрибуты
BachelorStudent bachelor = new BachelorStudent("Иван", 20, 3, "ТИУ", "Программная инженерия", 5000m)
    { Email = "ivan@tiu.ru", InternshipCompany = "Яндекс" };
MasterStudent master = new MasterStudent("Елена", 24, 1, "ТИУ", "Профессор Сидоров", 2)
    { Email = "elena@tiu.ru" };
PhDStudent phd = new PhDStudent("Александр", 28, 4, "ТИУ", "Машинное обучение в биоинформатике", "AI-Grant-2026", 15000m)
    { Email = "alexandr@tiu.ru" };
SeniorPhDStudent seniorPhd = new SeniorPhDStudent("Дмитрий", 30, 5, "ТИУ", "Квантовые вычисления", "Quantum-Grant", 20000m, true)
    { Email = "dmitry@tiu.ru" };

// 1. Простое наследование + новый метод
Console.WriteLine("--- 1. Простое наследование ---");
bachelor.AttendLecture();
bachelor.Study();
bachelor.ReceiveScholarship();
bachelor.DoInternship();      

// 2. Множественное наследование + новый метод
Console.WriteLine("\n--- 2. Множественное наследование (интерфейсы) ---");
phd.TeachClass();
phd.ReceiveScholarship();
phd.FilePatent();            

// 3. Сложное наследование
Console.WriteLine("\n--- 3. Сложное наследование ---");
seniorPhd.PrepareDefense();
seniorPhd.Study();
```

```
// 4. Взаимодействие объектов + новые методы
Console.WriteLine("\n--- 4. Взаимодействие объектов ---");
bachelor.AskForHelp(master);
master.Mentor(bachelor);
phd.ShareKnowledge(bachelor);
bachelor.SendEmail("Приглашение на консультацию"); // НОВЫЙ метод базового класса
master.AttendConference();                          // НОВЫЙ метод

// 5. Перекрытие (полиморфизм через базовый тип)
Console.WriteLine("\n--- 5. Перекрытие (override): полиморфный вызов GetInfo() ---");
Student[] university = { bachelor, master, phd, seniorPhd };
foreach (Student student in university)
{
    Console.WriteLine(student.GetInfo());
}

// 6. Перегрузка методов
Console.WriteLine("\n--- 6. Перегрузка (overload): один метод, разные параметры ---");
bachelor.Study();
bachelor.Study(6);
bachelor.Study("Математика");
master.TeachClass();
master.TeachClass("Физика");
```

```
// 7. Генерический класс
Console.WriteLine("--- 7. Generic класс StudentCollection<T> ---");
StudentCollection<Student> allStudents = new StudentCollection<Student>();
allStudents.Add(bachelor);
allStudents.Add(master);
allStudents.Add(phd);
allStudents.Add(seniorPhd);
Console.WriteLine($"Студентов: {allStudents.Count}");
allStudents.AllStudy();

// 8. НОВОЕ: ЯВНАЯ реализация интерфейса ICanPublish
Console.WriteLine("\n--- 8. Явная реализация интерфейса ICanPublish ---");
ICanPublish publisher1 = master; 
publisher1.PublishWork();        
ICanPublish publisher2 = phd;
publisher2.PublishWork();

// 9. НОВОЕ: УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (Dependency Injection)
Console.WriteLine("\n--- 9. Управление зависимостями (DI) ---");
ScholarshipService scholarshipService = new ScholarshipService(bachelor);
scholarshipService.PayScholarship();

ScholarshipService phdService = new ScholarshipService(phd); 
phdService.PayScholarship();

TeachingService teachingService = new TeachingService(master); 
teachingService.ScheduleClass();
```
